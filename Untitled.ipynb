```python
import pytesseract
from PIL import Image

pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

print(pytesseract.get_tesseract_version())
```

```python
from PIL import Image, ImageDraw, ImageFont

# Create a blank white image
img = Image.new("RGB", (400, 500), color="white")
draw = ImageDraw.Draw(img)

receipt_text = """
   CORNER MARKET
   123 Main St
   Rancho Cucamonga, CA

   Date: 09/18/2026

   Milk              3.49
   Bread             2.99
   Eggs              4.29
   Bananas           1.89
   Coffee            8.99

   Subtotal         21.65
   Tax               1.95
   TOTAL            23.60

   Thank you!
"""

draw.text((20, 20), receipt_text, fill="black")
img.save("sample_receipt.png")
img
```

```python
image = Image.open("sample_receipt.png")
raw_text = pytesseract.image_to_string(image)

print(raw_text)
```

```python
import re

lines = raw_text.split("\n")

items = []
for line in lines:
    # Look for a line ending in a number (with or without a decimal)
    match = re.search(r"([A-Za-z ]+)\s+(\d+\.?\d*)$", line.strip())
    if match:
        item_name = match.group(1).strip()
        price = match.group(2)
        items.append({"item": item_name, "price": price})

for i in items:
    print(i)
```

```python
def fix_price(price):
    price = price.replace(",", "")  # just in case
    if "." not in price:
        price = price[:-2] + "." + price[-2:]
    return price

for i in items:
    i["price"] = fix_price(i["price"])

for i in items:
    print(i)
```

```python
items = []
for line in lines:
    match = re.search(r"([A-Za-z]+)\s*[—\-]?\s*(\d+\.?\d*)$", line.strip())
    if match:
        item_name = match.group(1).strip()
        price = fix_price(match.group(2))
        items.append({"item": item_name, "price": price})

for i in items:
    print(i)
```

```python
corrections = {
    "Mik": "Milk",
    "Tex": "Tax"
}

for i in items:
    if i["item"] in corrections:
        i["item"] = corrections[i["item"]]

for i in items:
    print(i)
```

```python
import pandas as pd

df = pd.DataFrame(items)
df["price"] = df["price"].astype(float)

print(df)

df.to_csv("receipt_data.csv", index=False)
```

```python
item_total = df[~df["item"].isin(["Subtotal", "Tex", "Tax", "TOTAL"])]["price"].sum()
print(f"Sum of items: {item_total}")
```
